# Lab 02: THỰC HÀNH IMAGE FILTER

```python
import cv2
import matplotlib.pyplot as NguyenPhuocThanh
img = cv2.imread('cat.jpg')
img = img[:,:,::-1]
NguyenPhuocThanh.imshow(img)
```

1. Sử dụng boxFilter: cv2.blur() hoặc cv2.boxFilter() để làm mờ ảnh (sử dụng nhiều thông số filter khác nhau), biểu diễn ảnh gốc và ảnh làm mờ cùng nhau để kiểm chứng.

```python
blur1 = cv2.blur(img, (11,11))
blur2 = cv2.boxFilter(img, -1, (11,11))
blur3 = cv2.blur(img, (21,21))
blur4 = cv2.boxFilter(img, -1, (21,21))

NguyenPhuocThanh.figure(figsize=(15,10))

NguyenPhuocThanh.subplot(2,3,1)
NguyenPhuocThanh.imshow(img)
NguyenPhuocThanh.title("Original")

NguyenPhuocThanh.subplot(2,3,2)
NguyenPhuocThanh.imshow(blur1)
NguyenPhuocThanh.title("Blur 11x11")

NguyenPhuocThanh.subplot(2,3,3)
NguyenPhuocThanh.imshow(blur2)
NguyenPhuocThanh.title("BoxFilter 11x11")

NguyenPhuocThanh.subplot(2,3,4)
NguyenPhuocThanh.imshow(img)
NguyenPhuocThanh.title("Original")

NguyenPhuocThanh.subplot(2,3,5)
NguyenPhuocThanh.imshow(blur3)
NguyenPhuocThanh.title("Blur 21x21")

NguyenPhuocThanh.subplot(2,3,6)
NguyenPhuocThanh.imshow(blur4)
NguyenPhuocThanh.title("BoxFilter 21x21")

NguyenPhuocThanh.show()
```

2. Sử dụng Gaussian Filter # Lưu ý: kích thước bộ lọc là số lẻ.

```python
gauss1 = cv2.GaussianBlur(img, (7,7), 0)
gauss2 = cv2.GaussianBlur(img, (11, 11), 0)
gauss3 = cv2.GaussianBlur(img, (21,21), 0)

NguyenPhuocThanh.figure(figsize=(10,8))

NguyenPhuocThanh.subplot(2,2,1)
NguyenPhuocThanh.imshow(img)
NguyenPhuocThanh.title("original image")

NguyenPhuocThanh.subplot(2,2,2)
NguyenPhuocThanh.imshow(gauss1)
NguyenPhuocThanh.title("gaussian 7x7")

NguyenPhuocThanh.subplot(2,2,3)
NguyenPhuocThanh.imshow(gauss2)
NguyenPhuocThanh.title("gaussian 11x11")

NguyenPhuocThanh.subplot(2,2,4)
NguyenPhuocThanh.imshow(gauss3)
NguyenPhuocThanh.title("gaussian 21x21")

NguyenPhuocThanh.show()
```

3. Sử dụng Median Filter 

```python
med1 = cv2.medianBlur(img, 7)
med2 = cv2.medianBlur(img, 11)
med3 = cv2.medianBlur(img, 21)

NguyenPhuocThanh.figure(figsize=(10,8))

NguyenPhuocThanh.subplot(2,2,1)
NguyenPhuocThanh.imshow(img)
NguyenPhuocThanh.title("original image")

NguyenPhuocThanh.subplot(2,2,2)
NguyenPhuocThanh.imshow(med1)
NguyenPhuocThanh.title("median 7x7")

NguyenPhuocThanh.subplot(2,2,3)
NguyenPhuocThanh.imshow(med2)
NguyenPhuocThanh.title("median 11x11")

NguyenPhuocThanh.subplot(2,2,4)
NguyenPhuocThanh.imshow(med3)
NguyenPhuocThanh.title("median 21x21")

NguyenPhuocThanh.show()
```

4. Kiểm tra lại các bộ lọc trên với hai hình ảnh pepper noise dưới đây: 

```python
import numpy as np
noise1 = cv2.imread('pepper_noise01.jpg')
noise1 = noise1[:,:,::-1]
noise2 = cv2.imread('pepper_noise02.jpg')
noise2 = noise2[:,:,::-1]

def box_filter(image, ksize):
    return cv2.blur(image, (ksize, ksize))
def gaussian_filter(image, ksize):
    return cv2.GaussianBlur(image, (ksize, ksize), 0)
def median_filter(image, ksize):
    return cv2.medianBlur(image, ksize)
NguyenPhuocThanh.figure(figsize=(20,10))

NguyenPhuocThanh.subplot(2,4,1)
NguyenPhuocThanh.imshow(noise1)
NguyenPhuocThanh.title("pepper noise 1 original")

NguyenPhuocThanh.subplot(2,4,2)
NguyenPhuocThanh.imshow(box_filter(noise1, 5))
NguyenPhuocThanh.title("box filter")

NguyenPhuocThanh.subplot(2,4,3)
NguyenPhuocThanh.imshow(gaussian_filter(noise1, 5))
NguyenPhuocThanh.title("gaussian filter")

NguyenPhuocThanh.subplot(2,4,4)
NguyenPhuocThanh.imshow(median_filter(noise1, 5))
NguyenPhuocThanh.title("median filter")

NguyenPhuocThanh.subplot(2,4,5)
NguyenPhuocThanh.imshow(noise2)
NguyenPhuocThanh.title("pepper noise 2 original")

NguyenPhuocThanh.subplot(2,4,6)
NguyenPhuocThanh.imshow(box_filter(noise2, 5))
NguyenPhuocThanh.title("box filter")

NguyenPhuocThanh.subplot(2,4,7)
NguyenPhuocThanh.imshow(gaussian_filter(noise2, 5))
NguyenPhuocThanh.title("gaussian filter")

NguyenPhuocThanh.subplot(2,4,8)
NguyenPhuocThanh.imshow(median_filter(noise2, 5))
NguyenPhuocThanh.title("median filter")

NguyenPhuocThanh.show()
```

6. Thực hành với Cân bằng sáng - equalizeHist, để cân bằng sáng, trước hết ta chuyển ảnh trắng đen (nếu có) sang ảnh màu. 

```python
img = cv2.imread('pepper_noise02.jpg')
blur = cv2.medianBlur(img,5)
blur = cv2.cvtColor(blur, cv2.COLOR_RGB2GRAY)
equalized = cv2.equalizeHist(blur)

NguyenPhuocThanh.figure(figsize=(12,4))

NguyenPhuocThanh.subplot(1,2,1)
NguyenPhuocThanh.imshow(img)
NguyenPhuocThanh.title("Original")
NguyenPhuocThanh.axis('off')
NguyenPhuocThanh.subplot(1,2,2)
NguyenPhuocThanh.imshow(equalized, cmap='gray')
NguyenPhuocThanh.title("Equalized")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.show()
```

c. Process to have an image which has only the “ball”
Nếu vẫn còn xuất hiện các vạch trắng, cách khắc phục như thế nào?


```python
img = cv2.imread('exT3_01.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 50, 150)

# Xóa đường thẳng đứng (thanh) bằng opening kernel dọc
kernel = np.ones((25,3), np.uint8)
clean = cv2.morphologyEx(edges, cv2.MORPH_OPEN, kernel)

# Lấy phần còn lại = biên quả bóng
ball_edge = cv2.subtract(edges, clean)

NguyenPhuocThanh.imshow(ball_edge, cmap='gray')
NguyenPhuocThanh.title("Ball edge")
NguyenPhuocThanh.axis('off')
NguyenPhuocThanh.show()
```

Một số bài tập mở rộng: 
1. Sử dụng Backward, forward, central gradient để lọc ảnh. 


```python
img = cv2.imread('cat.jpg')
img = img[:,:,::-1]   
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

forward = np.array([[0, -1, 1]])
backward = np.array([[-1, 1, 0]])
central = np.array([[-1, 0, 1]]) / 2

fwd = cv2.filter2D(gray, -1, forward)
bwd = cv2.filter2D(gray, -1, backward)
cen = cv2.filter2D(gray, -1, central)

NguyenPhuocThanh.figure(figsize=(10,8))

NguyenPhuocThanh.subplot(2,2,1)
NguyenPhuocThanh.imshow(gray, cmap='gray')
NguyenPhuocThanh.title("Original")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(2,2,2)
NguyenPhuocThanh.imshow(fwd, cmap='gray')
NguyenPhuocThanh.title("Forward Gradient")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(2,2,3)
NguyenPhuocThanh.imshow(bwd, cmap='gray')
NguyenPhuocThanh.title("Backward Gradient")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(2,2,4)
NguyenPhuocThanh.imshow(cen, cmap='gray')
NguyenPhuocThanh.title("Central Gradient")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.show()
```

2. Sử dụng Finite difference filter để lọc ảnh nhiễu. 

```python
img = cv2.imread('cat.jpg')
img = img[:,:,::-1]
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

dx = np.array([[1, -1]])
dy = np.array([[1], [-1]])

gx = cv2.filter2D(gray, -1, dx)
gy = cv2.filter2D(gray, -1, dy)

grad = np.abs(gx) + np.abs(gy)

NguyenPhuocThanh.figure(figsize=(12,5))

NguyenPhuocThanh.subplot(1,3,1)
NguyenPhuocThanh.imshow(gray, cmap='gray')
NguyenPhuocThanh.title("Original")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(1,3,2)
NguyenPhuocThanh.imshow(gx, cmap='gray')
NguyenPhuocThanh.title("Finite Diff X")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(1,3,3)
NguyenPhuocThanh.imshow(grad, cmap='gray')
NguyenPhuocThanh.title("Finite Diff Magnitude")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.show()
```

3. Sử dụng Gaussian filter để lọc ảnh. 

```python
blur3 = cv2.GaussianBlur(img, (7,7), 0)
blur7 = cv2.GaussianBlur(img, (11,11), 0)
blur11 = cv2.GaussianBlur(img, (21,21), 0)

NguyenPhuocThanh.figure(figsize=(8,10))

NguyenPhuocThanh.subplot(2,2,1)
NguyenPhuocThanh.imshow(img)
NguyenPhuocThanh.title("Original")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(2,2,2)
NguyenPhuocThanh.imshow(blur3)
NguyenPhuocThanh.title("Gaussian 7x7")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(2,2,3)
NguyenPhuocThanh.imshow(blur7)
NguyenPhuocThanh.title("Gaussian 11x11")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(2,2,4)
NguyenPhuocThanh.imshow(blur11)
NguyenPhuocThanh.title("Gaussian 21x21")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.show()
```

4. Sử dụng Sobel detector để xác định biên của ảnh. 

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

sobel_mag = cv2.magnitude(sobel_x, sobel_y)

NguyenPhuocThanh.figure(figsize=(12,5))

NguyenPhuocThanh.subplot(1,4,1)
NguyenPhuocThanh.imshow(gray, cmap='gray')
NguyenPhuocThanh.title("Original")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(1,4,2)
NguyenPhuocThanh.imshow(np.abs(sobel_x), cmap='gray')
NguyenPhuocThanh.title("Sobel X")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(1,4,3)
NguyenPhuocThanh.imshow(np.abs(sobel_y), cmap='gray')
NguyenPhuocThanh.title("Sobel Y")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(1,4,4)
NguyenPhuocThanh.imshow(sobel_mag, cmap='gray')
NguyenPhuocThanh.title("Sobel Magnitude")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.show()
```

5. Sử dụng Canny edge detector để xác định các cạnh góc ảnh. 

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

edges = cv2.Canny(gray, 50, 150)

NguyenPhuocThanh.figure(figsize=(10,4))

NguyenPhuocThanh.subplot(1,2,1)
NguyenPhuocThanh.imshow(gray, cmap='gray')
NguyenPhuocThanh.title("Original")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(1,2,2)
NguyenPhuocThanh.imshow(edges, cmap='gray')
NguyenPhuocThanh.title("Canny Edges")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.show()
```

6. Sử dụng Hough transform để xác định các đường thẳng trong ảnh. 

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

edges = cv2.Canny(gray, 50, 150)

lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold=100,
                        minLineLength=50, maxLineGap=10)

img_lines = img.copy()
if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]
        cv2.line(img_lines, (x1, y1), (x2, y2), (255,0,0), 2)

NguyenPhuocThanh.figure(figsize=(10,4))

NguyenPhuocThanh.subplot(1,2,1)
NguyenPhuocThanh.imshow(edges, cmap='gray')
NguyenPhuocThanh.title("Edges")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.subplot(1,2,2)
NguyenPhuocThanh.imshow(img_lines)
NguyenPhuocThanh.title("Hough Lines")
NguyenPhuocThanh.axis('off')

NguyenPhuocThanh.show()
```
